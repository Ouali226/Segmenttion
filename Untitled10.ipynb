```python
# =============================================================================
# PROJET : SEGMENTATION CLIENTS (CLUSTERING)
# ENVIRONNEMENT : Python / Anaconda (Spyder)
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
import scipy.cluster.hierarchy as sch
from mpl_toolkits.mplot3d import Axes3D

# --- ÉTAPE 1 : CHARGEMENT ET EXPLORATION ---
df = pd.read_csv('Mall_Customers.csv')

# Visualisation des premières lignes et statistiques
print("Aperçu des données :")
print(df.head())

# Configuration esthétique des graphiques
sns.set(style="whitegrid")

# Graphique de distribution des variables clés
plt.figure(figsize=(16, 5))
plt.subplot(1, 3, 1)
sns.histplot(df['Age'], kde=True, color='blue')
plt.title('Distribution de l\'Âge')

plt.subplot(1, 3, 2)
sns.histplot(df['Annual Income (k$)'], kde=True, color='green')
plt.title('Distribution du Revenu Annuel')

plt.subplot(1, 3, 3)
sns.histplot(df['Spending Score (1-100)'], kde=True, color='red')
plt.title('Distribution du Score de Dépense')
plt.show()

# --- ÉTAPE 2 : PRÉPARATION DES DONNÉES ---
# Sélection des colonnes : Revenu et Score de dépense
X = df.iloc[:, [3, 4]].values

# Normalisation pour que les variables aient la même importance
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# --- ÉTAPE 3 : RECHERCHE DU NOMBRE OPTIMAL DE CLUSTERS ---
plt.figure(figsize=(14, 5))

# 1. Méthode du Coude (Elbow Method) pour K-Means
plt.subplot(1, 2, 1)
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Méthode du Coude')
plt.xlabel('Nombre de clusters')
plt.ylabel('WCSS')

# 2. Dendrogramme pour le Clustering Hiérarchique (CAH)
plt.subplot(1, 2, 2)
dendrogram = sch.dendrogram(sch.linkage(X_scaled, method='ward'))
plt.title('Dendrogramme (CAH)')
plt.xlabel('Clients')
plt.ylabel('Distances Euclidiennes')
plt.show()

# --- ÉTAPE 4 : APPLICATION DES ALGORITHMES (k=5) ---
# K-Means
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
df['Cluster_KMeans'] = kmeans.fit_predict(X_scaled)

# CAH
hc = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='ward')
df['Cluster_CAH'] = hc.fit_predict(X_scaled)

# --- ÉTAPE 5 : VISUALISATIONS COMPARATIVES ET MÉTIER ---

# 1. Visualisation des clusters (K-Means)
plt.figure(figsize=(10, 7))
sns.scatterplot(x='Annual Income (k$)', y='Spending Score (1-100)',
                hue='Cluster_KMeans', data=df, palette='bright', s=100, alpha=0.7)
plt.scatter(kmeans.cluster_centers_[:, 0] * scaler.scale_[0] + scaler.mean_[0],
            kmeans.cluster_centers_[:, 1] * scaler.scale_[1] + scaler.mean_[1],
            s=300, c='black', marker='X', label='Centroïdes')
plt.title('Segmentation Finale des Clients (K-Means)')
plt.legend()
plt.show()

# 2. Analyse du profil des segments (Boxplots)
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
sns.boxplot(x='Cluster_KMeans', y='Annual Income (k$)', data=df, hue='Cluster_KMeans', palette='Set2', legend=False)
plt.title('Niveau de Revenu par Segment')

plt.subplot(1, 2, 2)
sns.boxplot(x='Cluster_KMeans', y='Spending Score (1-100)', data=df, hue='Cluster_KMeans', palette='Set2', legend=False)
plt.title('Niveau de Dépense par Segment')
plt.show()

# --- ÉTAPE 6 : RÉSUMÉ DES SEGMENTS ---
summary = df.groupby('Cluster_KMeans')[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].mean()
print("\nCaractéristiques moyennes par segment :")
print(summary)
```

# Segmentation Client (Clustering)

## Introduction
Ce notebook a pour objectif de réaliser une segmentation client basée sur les données d'un centre commercial (`Mall_Customers.csv`). L'analyse utilise des techniques de clustering (K-Means et Clustering Hiérarchique) pour identifier des groupes de clients distincts en fonction de leurs caractéristiques démographiques et de leur comportement d'achat (Revenu Annuel et Score de Dépense).

## Étapes de l'Analyse

### 1. Chargement et Exploration des Données
*   Chargement du jeu de données `Mall_Customers.csv`.
*   Affichage des premières lignes et des statistiques descriptives.
*   Visualisation des distributions de l'âge, du revenu annuel et du score de dépense pour comprendre la répartition des données.

### 2. Préparation des Données
*   Sélection des variables clés pour la segmentation : `Annual Income (k$)` et `Spending Score (1-100)`.
*   Standardisation des données (`StandardScaler`) pour s'assurer que toutes les variables contribuent de manière égale à l'analyse de clustering.

### 3. Recherche du Nombre Optimal de Clusters
*   **Méthode du Coude (Elbow Method)** : Utilisée avec l'algorithme K-Means pour déterminer le nombre optimal de clusters en observant le point d'inflexion (le 'coude') de la courbe WCSS (Within-Cluster Sum of Squares).
*   **Dendrogramme (Clustering Hiérarchique)** : Un dendrogramme a été généré pour visualiser la structure hiérarchique des clusters et confirmer le nombre optimal de groupes.

Les deux méthodes ont suggéré un nombre optimal de 5 clusters.

### 4. Application des Algorithmes de Clustering
*   **K-Means** : L'algorithme K-Means a été appliqué avec 5 clusters. Les labels de cluster sont ajoutés au DataFrame original (`Cluster_KMeans`).
*   **Clustering Hiérarchique (CAH)** : L'algorithme `AgglomerativeClustering` a également été appliqué avec 5 clusters. Les labels de cluster sont ajoutés au DataFrame original (`Cluster_CAH`).

### 5. Visualisations Comparatives et Analyse Métier
*   **Visualisation des Clusters K-Means** : Un nuage de points montre les clients segmentés en fonction de leur revenu annuel et de leur score de dépense, avec les centroïdes des clusters mis en évidence.
*   **Analyse du Profil des Segments (Boxplots)** : Des boxplots sont utilisés pour visualiser la distribution du revenu annuel et du score de dépense pour chaque segment, permettant une meilleure compréhension des caractéristiques de chaque groupe.

### 6. Résumé des Segments
Un tableau récapitulatif a été créé, montrant les moyennes d'âge, de revenu annuel et de score de dépense pour chaque cluster.
